```python
def draw_bar_plot_for_tag(idx, df, option_dict, output_dir):
    sharey = False ## 또는 sharey=False
    sharex = True ## 또는 sharex=False

    fig, axs = plt.subplots(figsize = (12, 6), constrained_layout=True)
    alpha_list = [0.6, 0.5]
    color_list = ["#a6d96a", "#1a9850"]
    panel_list = ['A.', 'B.', 'C.', 'D.']


    x = list(df['rel_week'])
    y = list(df['count'])
    plotgen.draw_tag_bar(axs
                        , f'Number of questions for {option_dict["selected_tags"]}' 
                        , x
                        , y
                        , color_list[0]
                        , alpha_list[0]
                        , ''
                        , set_ylim = False
                        )
        

    fig.supxlabel("Weeks relative to ChatGPT release", fontsize=22) 
    fig.supylabel('Number of questions', fontsize=22)
    filename = f"{idx}_Countbar_plot_{option_dict['selected_tags']}_{option_dict['year_range']}.png"
    plt.savefig(
    os.path.join(output_dir, filename),
    dpi=300,
    bbox_inches='tight'
    )
    plt.close(fig)
    
```

```python
for idx, lang in enumerate(CONSTANTS.languages_from2020to2022) :
    print(f'[Start....] drawing figures for {lang} language')

    std_date = Date_Setting[option_dict['year_range']]['std_date']
    data_dir = f"{path_list['data_root_dir']}/data/{option_dict['snapshot']}/questions/{lang}/{option_dict['year_range']}/"
    option_dict['selected_tags'] = lang
    output_dir = './fig/'
    print(data_dir)
    df = load_df(data_dir, ['id', 'creationdate', 'title', 'tags', 'body'])
    df['creationdate'] = pd.to_datetime(df['creationdate'], format="mixed").dt.normalize()
    df['rel_week'] = np.floor((df['creationdate']-std_date).dt.days/7)
    tot_cnt = (
        df.groupby('rel_week')['id']
        .count()
        .reset_index(name='count')
    )
    draw_bar_plot_for_tag(idx, tot_cnt, option_dict, output_dir)
    print(f'[End....] drawing figures for {lang} language')


```
