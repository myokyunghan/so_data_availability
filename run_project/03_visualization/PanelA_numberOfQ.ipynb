```python
def compute_N_by_week(df, period_col='rel_week', weight_col='tot_pct'):
    """주차별 질문 수 (tot_pct를 질문 수로 쓰는 경우)"""
    N = df.groupby(period_col)[weight_col].sum()
    try:
        N.index = N.index.astype(int)
    except (ValueError, TypeError):
        pass
    return N.sort_index()

def plot_panelA(N_series,
                             reference_point=0, reference_label='ChatGPT launch',
                             figsize=(12, 6), save_path=None, lang = 'python'):
    """
    4-panel 그림:
    (A) 질문 수 N 시계열 — 감소 추세 확인
    (B) Baseline Top-30 share — 주류 태그 밀려남
    (C) Long-tail share — 롱테일 부상
    (D) Gini: raw vs N-adjusted — N 통제해도 변화 있음
    """
    fig, axes = plt.subplots(1, 1, figsize=figsize, sharex=True)

    # (A) N
    idx = N_series.index.values
    axes.bar(idx, N_series.values, color='gray', alpha=0.5, width=1.0,
                label='Questions per week')
    axes.set_ylabel('N (questions)', fontsize=11)
    axes.set_title(f'{lang}', fontsize=11, loc='left')
    axes.grid(alpha=0.3, axis='y')

    if reference_point is not None:
        axes.axvline(x=reference_point, color='red', linestyle='--',
                    linewidth=1, alpha=0.6)

    # x축
    all_idx = N_series.index.values
    tick_step = 26
    xmin = int(np.floor(all_idx.min() / tick_step) * tick_step)
    xmax = int(np.ceil(all_idx.max() / tick_step) * tick_step)
    axes.set_xticks(np.arange(xmin, xmax + 1, tick_step))
    axes.set_xlabel(f'Week (relative to {reference_label.lower()}, week 0)',
                        fontsize=11)

    fig.suptitle('Evidence for topic diversification despite declining volume',
                 fontsize=13, y=0.995, fontweight='bold')
    fig.tight_layout()
    if save_path:
        fig.savefig(save_path, dpi=150, bbox_inches='tight')
    plt.close()


```

```python
for idx, lang in enumerate(CONSTANTS.languages_from2020to2022) :
    plotgen = PlotGen(idx, 'tag')
    print(f'[Start....] drawing figures for {lang} language')


    df = load_df(plotgen.data_dir, ['cdate' , 'id' , 'tag', 'cnt', 'tot_cnt', 'pct'])
    df = calc_rel_period(df, plotgen.std_date, date_col = 'cdate', period = 7)
    # df_proportion = compute_proportion_period(df,period = 'rel_week', type = 'tag', value_col='pct')

    N_series = compute_N_by_week(df, weight_col='pct')

    # 종합 시각화
    plot_panelA(N_series, reference_point=0,
                             save_path=f'./fig/{idx}_PanelA_for_{lang}.png', lang = lang)


    print("\nSaved figures.")
```
